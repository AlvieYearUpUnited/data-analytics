# Exercise 2.C: Creating and calling functions, Lab 3

This exercise goes over creating functions to help calculate total tax amount. 

```python
def get_soc_sec_tax(gross_pay):
    tax_rate = 0.062
    ss_tax = gross_pay * tax_rate

    print(f'Gross Pay:                   ${gross_pay:,.2f}')
    print(f'Social Security Tax Rate:         {tax_rate * 100:,.2f}%')
    print(f'Social Security Tax Amount:     ${ss_tax:,.2f}')
    return ss_tax
    
get_soc_sec_tax(15000)
```

```python
def get_medicare_tax(gross_pay_amount):
    tax_rate = 0.0145
    medicare_tax = gross_pay_amount * tax_rate

    print(f'Gross Pay Amount: ${gross_pay_amount:,.2f}')
    print(f'Medicare Tax Rate: {tax_rate * 100:,.2f}')
    print(f'Medicare Tax Amount: ${medicare_tax:,.2f}')

get_medicare_tax(3953.34)
```

```python
def get_federal_tax(gross_pay_amount, withholding_code):
    if withholding_code == 0:
        tax_rate = .23    
    elif withholding_code == 1:
        tax_rate = .21
    elif withholding_code == 2:
        tax_rate = .195
    elif withholding_code == 3:
        tax_rate = .185
    elif withholding_code >= 4:
        tax_rate = .18 - .005 * (withholding_code - 4)
    else:
        print(f'ERROR!')

    fed_tax_withholding = gross_pay_amount * tax_rate

    print(f'Gross Pay Amount:         ${gross_pay_amount:,.2f}')
    print(f'Withholding Code:                 {withholding_code}')
    print(f'Federal Tax Rate:            {tax_rate * 100:,.2f}%')
    print(f'Federal Tax Witholding:     ${fed_tax_withholding:,.2f}')

person1 = (750, 0)
person2 = (1550, 2)
person3 = (1100, 5)

get_federal_tax(*person3)
```
