```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
day_of_week = today.strftime("%A")
```

```python
month = today.strftime("%B")
```

```python
day = today.strftime("%d")
```

```python
year = today.strftime("%Y")
```

```python
print(f'Today is {day_of_week}, {month} {day}, {year}')
```

```python
# My birthday
```

```python
my_bday = datetime.datetime(2000, 12, 11)
```

```python
bday_month = my_bday.strftime("%m")
```

```python
bday_day = my_bday.strftime("%d")
```

```python
bday_year = my_bday.strftime("%Y")
```

```python
print(f'My birthday is {bday_month}/{bday_day}/{bday_year}')
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days = 90)
```

```python
ninety_month = ninety_d.strftime("%B")
```

```python
ninety_day = ninety_d.strftime("%d")
```

```python
ninety_year = ninety_d.strftime("%Y")
```

```python
print(f'90 days from today is {ninety_month} {ninety_day}, {ninety_year}')
```
