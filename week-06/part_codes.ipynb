## Lab 2. Parsing strings - part codes

```python
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'
```

```python
def get_supplier_code(part):
    for i in range(len(part)):
        if part[i] == ':':
            supplier_code = part[:i]
            return supplier_code
```

```python
def get_product_num(part):
    seperator = []
    for i in range(len(part)):
        if part[i] == ':' or part[i] == '-':
            seperator.append(i)
    product_num = part[seperator[0] + 1:seperator[1]]
    return product_num
```

```python
def get_size(part):
    for i in range(len(part)):
        if part[i] == '-':
            size = part[i+1:]
            return size
```

```python
print(f'For {part1}, the supplier is {get_supplier_code(part1)}. Part number {get_product_num(part1)} is available in size {get_size(part1)}.')
```

```python
print(f'For {part2}, the supplier is {get_supplier_code(part2)}. Part number {get_product_num(part2)} is available in size {get_size(part2)}.')
```

```python
print(f'For {part3}, the supplier is {get_supplier_code(part3)}. Part number {get_product_num(part3)} is available in size {get_size(part3)}.')
```

```python
parts = [part1, part2, part3]
```

```python
for part in parts:
    print(f'For {part}, the supplier is {get_supplier_code(part)}. Part number {get_product_num(part)} is available in size {get_size(part)}.')
```
