## <i>Array Basics</i>

```python
import numpy as np
```

```python
fruits = ['apple', 'banana', 'cherry', 'date']
```

```python
arr_1 = np.array(fruits)
```

```python
print(arr_1)
```

```python
arr_1.dtype # U is for Unicode strings stored, 6 is for the max string length in this array
```

```python
arr_2 = np.array([
    [10, 11, 12],
    [13, 14, 15]
])
```

```python
print(arr_2)
```

```python
print(arr_2.T) # arr_2.T temporarily swaps rows and columns for this output 
```

```python
arr_3 = np.array([
    [1,3,5],
    [2,4,6]
])
```

```python
print(arr_3)
```

```python
arr_4 = np.array(
    [arr_3[0,2],
     arr_3[1,1]
     ]

)
```

```python
print(arr_4)
```

```python
arr_5 = np.ones((2, 2, 2))  # (layers, rows, columns)
```

```python
print(arr_5)
```
