```python
def favorite_things(name, movie, music):
    return f'Name:              {name} \nFavorite Movie:    {movie} \nFavorite Musician: {music}'
print(favorite_things('Alvie', 'Avengers', 'Drake'))
```

```python
def why_im_here(purpose):
    return f'My Purpose: {purpose}'
print(why_im_here('I joined the Year Up United Data Analyst program to gain technical skills and begin a career in Data Analytics.'))
```

```python
def favorite_place(place, why):
    return f'One of my favorite places is {place} because {why}.'
```

```python
print(favorite_place('New York', 'it is nice'))
```

```python
print(favorite_place('Florida', 'it is hot'))
```
