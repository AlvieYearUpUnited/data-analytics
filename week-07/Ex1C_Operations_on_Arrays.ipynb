## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

Rows were randomly shuffled, but its columns stayed in the same order.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

I would use shuffle() when I don't need to retain original order, permutation() when I do.

```python
print(r_2)
print(r_2.shape)
print(r_4)
print(r_4.shape)
print(np.add(r_2, r_4))       # ValueError occurs because the arrays have incompatible column sizes
print(np.add(r_2, r_4).shape) # Columns must match, or one of them must be 1
```

```python
np.add(r_4, np.array([1]))
```

```python
np.add(r_4, np.array([[1]]))
```

```python
np.add(r_4, np.array([1, 1, 1, 1, 1]))
```

```python
np.add(r_4, np.array([[1], [1]]))
```

When combining the values of two arrays, broadcasting requires dimension sizes (rows and columns) to match or one of them to be 1. Values are added together based on the matching index position. The resulting array's shape is determined by the larger size in each dimension.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print(bag1)
print(bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)
print(f'The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items')
```

```python
total = np.add(bag1, bag2)
print(total)
```

```python
print(f'There are {total[0]} vegetables, {total[1]} fruits, {total[2]} boxes of snacks, and {total[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])
```

```python
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.stack((ali, ben, carla))
print(all_knocks)
```

```python
print(f'Ali: {np.sum(all_knocks, axis = 1)[0]} total knocks')
print(f'Ben: {np.sum(all_knocks, axis = 1)[1]} total knocks')
print(f'Carla: {np.sum(all_knocks, axis = 1)[2]} total knocks')
print(f'Group: {np.sum(all_knocks)} total knocks')
```

```python
daily_knocks = np.sum(all_knocks, axis = 0) # Total group knocks per day
print(daily_knocks)
print()
rolling_knocks = np.cumsum(daily_knocks)    # Rolling total group knocks per day
print(rolling_knocks)
print()
estimated_votes = rolling_knocks * .5      # Estimated one vote per two knocks
print(estimated_votes)
```
