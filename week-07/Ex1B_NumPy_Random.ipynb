## Generating Random Choices from an Array

```python
students = np.array(['Joshua', 'Jake', 'Alex', 'Miguel', 'Mike', 'Joe', 'Kevin', 'John', 'Peter', 'Blake', 'Griffin', 'Stewie'])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size = (3, 4))
print(breakout_groups) # Not effective because students can be chosen more than once or not at all
```

```python
call_on_weighted = np.random.choice(students, size = 5,
                                    p = [.2, .3, .1, .05, .04, .01, .1, .1, .02, .04, .04, .0])
print(call_on_weighted) # p values are assigned based on index position. Jake has highest probability at 30%, Stewie has lowest at 0%
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(10, 5, size = (4, 5))
print(bell_curve)
```
