# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([
    (11, 12, 13),
    (14, 15, 16)
])
print(arrA)
    
```

```python
for n in arrA: 
    print(n * 2)
```

```python
for n in arrA:
    for i in n:
        print(i * 2)
```

First dimension: Prints each individual row per loop, multiplied by 2  
Second dimension: Printed each individual value from entire array, multiplied by 2

```python
for n in arrA:
    for i in n:
        loop_result = n * 2
print(loop_result)
```

'n in arrA': Looping through entire array, each row being an individual element

'i in n': Looping through each individual values in current row

'loop_result' variable's is being reassigned a value during each loop.
Since print statement is outside of all loops, it only prints the last value that loop_result holds

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
```

```python
print(f'joined_array_1 SHAPE {joined_array_1.shape}')
print(f'joined_array_2 SHAPE {joined_array_2.shape}')
```

ValueError recieved because np.stack requires all arrays to have the same shape

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
print(another_2)
```

ValueError recieved because .concatenate(), axis = 1 requires rows to be equal. arrA has two rows, joined_array_1 has four.   

```python
print(joined_array_1.shape)
print(arrA.shape)
```

```python
hstack_array = np.hstack((arrA, arrA))
print(hstack_array) # hstack adds columns
```

```python
vstack_array = np.vstack((arrA, arrA))
print(vstack_array) # vstack adds rows
```

```python
dstack_array = np.dstack((arrA, arrA))
print(dstack_array) # dstack adds layers
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky) # Two 2D arrays
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino) # Three 2D arrays: double brackets used
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
print(arrB)
```

```python
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print()
print(arrC)
```

```python
print(arrB)
y = arrB >= 7
arrD = arrB[y]
print()
print(arrD)
```
